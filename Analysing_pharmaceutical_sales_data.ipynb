# **What are the total sales quantities for each drug category (ATC code)?**

**Total M01AB Sales**

```python
total_m01ab_sales = float(sales["M01AB"].sum().round(2))
total_m01ab_sales
```

**Total M01AE Sales**

```python
total_m01ae_sales = float(sales["M01AE"].sum().round(2))
total_m01ae_sales
```

**Total N02BA Sales**

```python
total_n02ba_sales = float(sales["N02BA"].sum().round(2))
total_n02ba_sales
```

**Total N02BE Sales**

```python
total_n02be_sales = float(sales["N02BE"].sum().round(2))
total_n02be_sales
```

**Total N05B Sales**

```python
total_n05b_sales = float(sales["N05B"].sum().round(2))
total_n05b_sales
```

**Total N05C Sales**

```python
total_n05c_sales = float(sales["N05C"].sum().round(2))
total_n05c_sales
```

**Total R03 Sales**

```python
total_ro3_sales = float(sales["R03"].sum().round(2))
total_ro3_sales
```

**Total R06 Sales**

```python
total_r06_sales = float(sales["R06"].sum().round(2))
total_r06_sales
```

# **Which individual drug brands have the highest total sales?**

**Putting total sales of each drug in a dictionary**

```python
total_sales = {
          "M01AB": total_m01ab_sales,
          "M01AE": total_m01ae_sales,
          "N02BA": total_n02ba_sales,
          "N02BE": total_n02be_sales,
          "N05B": total_n05b_sales,
          "N05C": total_n05c_sales,
          "R03":total_ro3_sales,
          "R06": total_r06_sales
          }

drug_with_highest_sales = max(total_sales, key=total_sales.get)
highest_sales_value = total_sales[drug_with_highest_sales]
```

**Highest Sale**

```python
print(f"Drug {drug_with_highest_sales} has the highest sales of {highest_sales_value}")
```

# **Which three drugs have the highest sales in January 2015, July 2016, September 2017.**

# ***Which three drugs have the highest sales in January 2015.***

**Putting them all in a dict and using sorted()**

```python
jan_2015_sales = {
          "M01AB": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["M01AB"].sum().round(2)),
          "M01AE": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["M01AE"].sum().round(2)),
          "N02BA": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["N02BA"].sum().round(2)),
          "N02BE": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["N02BE"].sum().round(2)),
          "N05B": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["N05B"].sum().round(2)),
          "N05C": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["N05C"].sum().round(2)),
          "R03": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["R03"].sum().round(2)),
          "R06": float(sales[((sales["Year"] == 2015) & (sales["Month"] == 1))]["R06"].sum().round(2))
          }
```

```python
sorted_jan_2015_sales = sorted_sales = sorted(jan_2015_sales.items(), key=lambda x: x[1], reverse=True)
sorted_jan_2015_sales
```

**Three drugs which have highest sales during January 2015**

```python
print(f"Three drugs which have highest sales during January 2015")
for index, drug in enumerate(sorted_jan_2015_sales[:3], start=1):
  print(f"{index}, {drug[0]}")
```
